###### Basics

```python
import pandas as pd
# Creatinf a Series 
empty_series = pd.Series(data = None, dtype='int')
print(empty_series)
type(empty_series)
```

```python
# Creating a custom series with index and title

production_date = ['1001','1002','1003','1004','1005']
production_day =['Day01','Day02','Day03','Day04','Day05']

daily_production = pd.Series(data=production_day,index=production_date,name='Production Details')
print(daily_production)

print(daily_production.name) # printing the name of the series

# the length of the index and data has to be same otherwise it will give an error.Since the pandas is build on top of Numpy, it has default ndex stating with 0.


```

##### Some important Operation with the Series in Pandas


```python
month = ["Jan","Feb","Mar","Apr","May","Jun"]
sale =["$100","$500","$400","$700","$999","$499"]

sales_survay = pd.Series(data=sale,index=month,name="Sale Survey")
print(sales_survay)

# Give the name of the Series
print(sales_survay.name)

# give the index
print(sales_survay.index)

# give the values or data
print(sales_survay.values)

# gives the data type of the series
print(sales_survay.dtype)

# gives the shape of the series
print(sales_survay.shape)

# gives the dimention of the series
print(sales_survay.ndim)

# gives the length of the series
print(sales_survay.size)

# also give the length
len(sales_survay)

# checking empty or not
print(sales_survay.empty)
```
